````python
# ============================================================
# 第 1 格：安裝套件、讀取金鑰、設定模型（請勿修改）
# ============================================================
!pip install -q groq gradio

import os
import re
import time
import sqlite3
import pandas as pd
from IPython.display import display
from groq import Groq
import gradio as gr

# 判斷目前是在 Colab 還是本地端 Jupyter
try:
    from google.colab import userdata
    IN_COLAB = True
except ImportError:
    IN_COLAB = False

# 🔴 金鑰名稱一律為 GROQ_API_KEY；絕對不可以把金鑰寫在程式裡，也不可以 print 出來
if IN_COLAB:
    try:
        GROQ_API_KEY = userdata.get('GROQ_API_KEY')
    except Exception:
        print("❌ 讀不到金鑰：請到左側 🔑 新增名稱為 GROQ_API_KEY 的密鑰，並打開「筆記本存取權」")
        raise
else:
    GROQ_API_KEY = os.environ.get('GROQ_API_KEY')
    if not GROQ_API_KEY:
        raise RuntimeError("找不到環境變數 GROQ_API_KEY")

client = Groq(api_key=GROQ_API_KEY)

# 期中考指定模型；推理強度設為 low，可大幅減少 token 用量
# （唯一允許修改處：若 20b 額度用完，可改為 "openai/gpt-oss-120b"，並在規格書最後註明）
MODEL = "openai/gpt-oss-20b"
REASONING_EFFORT = "low"
print("✅ 環境：", "Colab" if IN_COLAB else "本地端", "｜模型：", MODEL)
````

````python
# ============================================================
# 第 2 格：建立資料庫與初始資料
# 提醒：每次執行都要 DROP TABLE IF EXISTS 再重建
# =============================================================
import sqlite3

conn = sqlite3.connect('activity_system.db')
cursor = conn.cursor()

try:
    cursor.execute("DROP TABLE IF EXISTS registrations")
    cursor.execute("DROP TABLE IF EXISTS activities")

    cursor.execute("""
        CREATE TABLE activities (
            id INTEGER PRIMARY KEY AUTOINCREMENT,
            name TEXT NOT NULL,
            date TEXT NOT NULL,
            capacity INTEGER NOT NULL
        )
    """)

    cursor.execute("""
        CREATE TABLE registrations (
            id INTEGER PRIMARY KEY AUTOINCREMENT,
            student_name TEXT NOT NULL,
            activity_name TEXT NOT NULL,
            count INTEGER NOT NULL,
            status TEXT NOT NULL
        )
    """)

    initial_activities = [
        ('迎新晚會', '2026-11-20', 6),
        ('程式工作坊', '2026-11-25', 4),
        ('籃球友誼賽', '2026-11-28', 3)
    ]

    cursor.executemany("""
        INSERT INTO activities (name, date, capacity)
        VALUES (?, ?, ?)
    """, initial_activities)

    conn.commit()
    print("✅ 資料庫建立成功，並已載入初始資料！")

except Exception as e:
    conn.rollback()
    print(f"❌ 建立資料庫時發生錯誤：{e}")

finally:
    conn.close()
````

````python
# ============================================================
# 第 3 格：顯示資料表
# 提醒：定義 show_tables() 並呼叫一次
# ============================================================
import sqlite3
import pandas as pd
from IPython.display import display

def show_tables():
    conn = sqlite3.connect('activity_system.db')
    try:
        print("📋 【活動表 (activities)】")
        df_activities = pd.read_sql_query("SELECT * FROM activities", conn)
        display(df_activities)

        print("\n📋 【報名紀錄表 (registrations)】")
        df_registrations = pd.read_sql_query("SELECT * FROM registrations", conn)
        display(df_registrations)

    finally:
        conn.close()

show_tables()
````

````python
# ============================================================
# 第 4 格：AI 解讀函式
# 提醒：回傳 (動作, 活動, 人數, 原始回覆)；不可直接相信 AI 的回覆
# ============================================================
import time
import re

ai_cache = {}

def parse_user_input(text):
    if text in ai_cache:
        return ai_cache[text]

    valid_actions = ["報名", "取消", "查詢", "統計"]
    valid_activities = ["迎新晚會", "程式工作坊", "籃球友誼賽"]

    prompt = f"""
你是一個系學會活動系統的 AI 助理。請從使用者的話語中抽取出三個資訊：動作、活動、人數。
規則：
1. 動作只能從這裡選一個：報名, 取消, 查詢, 統計。如果都不是，請填「未知」。
2. 活動只能從這裡選一個：迎新晚會, 程式工作坊, 籃球友誼賽。如果都不是，請填「未知」。
3. 人數請填入整數（包含自己和攜伴，例如「我和兩個朋友」代表 3 人）。如果沒有提到人數，請預設填 1。
4. 輸出格式必須嚴格為：動作,活動,人數
例如：「我要報名迎新晚會，我和兩個朋友」要輸出「報名,迎新晚會,3」

使用者說的話是："{text}"
請直接輸出格式結果，不要有其他多餘文字。
"""

    raw_response = ""
    for attempt in range(3):
        try:
            response = client.chat.completions.create(
                model=MODEL,
                messages=[{"role": "user", "content": prompt}],
                temperature=0,
                reasoning_effort=REASONING_EFFORT
            )
            raw_response = response.choices[0].message.content.strip()
            time.sleep(2)
            break
        except Exception as e:
            print(f"⚠️ 呼叫 AI 發生錯誤 (嘗試 {attempt+1}/3): {e}")
            time.sleep(30)
            if attempt == 2:
                raw_response = "未知,未知,1"

    action = "未知"
    activity = "未知"
    count = 1

    try:
        parts = raw_response.split(",")
        if len(parts) >= 3:
            extracted_action = parts[0].strip()
            extracted_activity = parts[1].strip()
            extracted_count_str = parts[2].strip()

            if extracted_action in valid_actions:
                action = extracted_action
            if extracted_activity in valid_activities:
                activity = extracted_activity

            numbers = re.findall(r'\d+', extracted_count_str)
            if numbers:
                parsed_num = int(numbers[0])
                if 1 <= parsed_num <= 3:
                    count = parsed_num
        else:
            for act in valid_actions:
                if act in text:
                    action = act
            for actv in valid_activities:
                if actv in text:
                    activity = actv
            numbers = re.findall(r'\d+', text)
            if numbers:
                parsed_num = int(numbers[0])
                if 1 <= parsed_num <= 3:
                    count = parsed_num
    except Exception as e:
        print(f"⚠️ 解析 AI 回覆時發生例外：{e}")

    result = (action, activity, count, raw_response)
    ai_cache[text] = result
    return result

print("✅ 第 4 格載入完成！")
````

````python
# ============================================================
# 第 5 格：AI 解讀測試
# 提醒：8～10 句，要與規格書 7-1 一致
# ============================================================
test_cases = [
    ("我要報名迎新晚會，我和兩個朋友", "報名", "迎新晚會", 3),
    ("查詢程式工作坊剩餘名額", "查詢", "程式工作坊", 1),
    ("我要取消迎新晚會", "取消", "迎新晚會", 1),
    ("各活動報名狀況", "統計", "未知", 1),
    ("幫我報名籃球友誼賽", "報名", "籃球友誼賽", 1),
    ("購買明天的高鐵票", "未知", "未知", 1),
    ("我要報名程式工作坊，我們 2 個人", "報名", "程式工作坊", 2),
    ("請問迎新晚會還有位子嗎", "查詢", "迎新晚會", 1)
]

passed_count = 0
print("🧪開始執行 AI 解讀測試...\n" + "="*50)

for i, (text, exp_action, exp_activity, exp_count) in enumerate(test_cases, 1):
    action, activity, count, raw = parse_user_input(text)

    is_pass = (action == exp_action) and (activity == exp_activity) and (count == exp_count)
    if is_pass:
        passed_count += 1
        status_str = "✅ 通過"
    else:
        status_str = "❌ 失敗"

    print(f"測試 {i}: {status_str}")
    print(f"  💬 使用者說：{text}")
    print(f"  🤖 AI 原始回覆：{raw}")
    print(f"  🔍 系統解讀結果：動作={action}, 活動={activity}, 人數={count}")
    print(f"  🎯 預期結果：動作={exp_action}, 活動={exp_activity}, 人數={exp_count}")
    print("-" * 50)

print(f"\n📊 測試總結：共 {len(test_cases)} 句，成功通過 {passed_count} 句")
````

````python
# ============================================================
# 第 6 格：功能函式
# 提醒：fetchone() 後先檢查 None；用 try/finally 關閉連線；統計用 JOIN＋GROUP BY
# ============================================================
import sqlite3
import pandas as pd
from IPython.display import display

def register_activity(student_name, activity_name, count):
    conn = sqlite3.connect('activity_system.db')
    try:
        cursor = conn.cursor()

        cursor.execute("SELECT capacity FROM activities WHERE name = ?", (activity_name,))
        row = cursor.fetchone()
        if row is None:
            return f"❌ 找不到活動：「{activity_name}」"
        capacity = row[0]

        cursor.execute("SELECT COUNT(*) FROM registrations WHERE student_name = ? AND activity_name = ? AND status = '已報名'", (student_name, activity_name))
        dup_row = cursor.fetchone()
        if dup_row and dup_row[0] > 0:
            return f"❌ {student_name} 已經報名過「{activity_name}」了，不可重複報名。"

        cursor.execute("SELECT COUNT(DISTINCT activity_name) FROM registrations WHERE student_name = ? AND status = '已報名'", (student_name,))
        active_count_row = cursor.fetchone()
        active_acts = active_count_row[0] if active_count_row else 0
        if active_acts >= 2:
            return f"❌ {student_name} 已同時報名 2 個活動，達到個人上限。"

        cursor.execute("SELECT SUM(count) FROM registrations WHERE activity_name = ? AND status = '已報名'", (activity_name,))
        sum_row = cursor.fetchone()
        current_booked = sum_row[0] if sum_row and sum_row[0] is not None else 0

        if current_booked + count > capacity:
            return f"❌ 報名失敗，「{activity_name}」名額不足（剩餘名額：{capacity - current_booked}）"

        cursor.execute("""
            INSERT INTO registrations (student_name, activity_name, count, status)
            VALUES (?, ?, ?, '已報名')
        """, (student_name, activity_name, count))
        conn.commit()
        return f"✅ {student_name} 成功報名「{activity_name}」，人數：{count}。"
    finally:
        conn.close()

def cancel_activity(student_name, activity_name):
    conn = sqlite3.connect('activity_system.db')
    try:
        cursor = conn.cursor()
        cursor.execute("""
            SELECT id FROM registrations
            WHERE student_name = ? AND activity_name = ? AND status = '已報名'
        """, (student_name, activity_name))
        row = cursor.fetchone()
        if row is None:
            return f"❌ 找不到 {student_name} 針對「{activity_name}」的有效報名紀錄。"

        reg_id = row[0]
        cursor.execute("UPDATE registrations SET status = '已取消' WHERE id = ?", (reg_id,))
        conn.commit()
        return f"✅ 已成功取消 {student_name} 對「{activity_name}」的報名。"
    finally:
        conn.close()

def query_activity(activity_name):
    conn = sqlite3.connect('activity_system.db')
    try:
        cursor = conn.cursor()
        cursor.execute("SELECT date, capacity FROM activities WHERE name = ?", (activity_name,))
        row = cursor.fetchone()
        if row is None:
            return f"❌ 找不到活動：「{activity_name}」"
        date, capacity = row

        cursor.execute("SELECT SUM(count) FROM registrations WHERE activity_name = ? AND status = '已報名'", (activity_name,))
        sum_row = cursor.fetchone()
        booked = sum_row[0] if sum_row and sum_row[0] is not None else 0
        remaining = capacity - booked
        return f"📌 活動名稱：{activity_name} | 日期：{date} | 總名額：{capacity} | 已報名人數：{booked} | 剩餘名額：{remaining}"
    finally:
        conn.close()

def get_statistics():
    conn = sqlite3.connect('activity_system.db')
    try:
        query = """
            SELECT
                a.name AS 活動名稱,
                a.date AS 日期,
                a.capacity AS 總名額,
                COALESCE(SUM(CASE WHEN r.status = '已報名' THEN r.count ELSE 0 END), 0) AS 已報名人數,
                a.capacity - COALESCE(SUM(CASE WHEN r.status = '已報名' THEN r.count ELSE 0 END), 0) AS 剩餘名額
            FROM activities a
            LEFT JOIN registrations r ON a.name = r.activity_name
            GROUP BY a.id, a.name, a.date, a.capacity
        """
        df = pd.read_sql_query(query, conn)
        display(df)
        return "📊 各活動報名狀況統計如下："
    finally:
        conn.close()

print("✅ 第 6 格載入完成！")
````

````python
# ============================================================
# 第 7 格：主流程與流程測試
# 提醒：流程測試 8～10 步；最後呼叫 show_tables()
# ============================================================
def process_user_request(student_name, text):
    action, activity, count, raw = parse_user_input(text)

    if action == "報名":
        if activity == "未知":
            return "❌ 無法識別您想報名哪一個活動。"
        return register_activity(student_name, activity, count)
    elif action == "取消":
        if activity == "未知":
            return "❌ 無法識別您想取消哪一個活動。"
        return cancel_activity(student_name, activity)
    elif action == "查詢":
        if activity == "未知":
            return "❌ 無法識別您想查詢哪一個活動。"
        return query_activity(activity)
    elif action == "統計":
        return get_statistics()
    else:
        return f"🤖 AI 回覆：目前有的活動為：迎新晚會、程式工作坊、籃球友誼賽。請重新輸入您的需求。"

workflow_test_cases = [
    ("小明", "查詢迎新晚會剩餘名額"),
    ("小明", "我要報名迎新晚會，我和兩個朋友"),
    ("小明", "查詢迎新晚會剩餘名額"),
    ("小華", "我要報名迎新晚會，我們 4 個人"),
    ("小華", "我要報名程式工作坊"),
    ("小華", "查詢程式工作坊剩餘名額"),
    ("小明", "各活動報名狀況"),
    ("小明", "我要取消迎新晚會")
]

print("🧪 開始執行流程測試...\n" + "="*50)
for i, (name, text) in enumerate(workflow_test_cases, 1):
    print(f"步驟 {i} | 姓名：{name} | 說：{text}")
    res = process_user_request(name, text)
    print(f"系統回應：\n{res}")
    print("-" * 50)

print("\n📋 流程測試結束後的最終資料表狀態：")
show_tables()
````
